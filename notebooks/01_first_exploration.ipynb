# First Iris exploration
The original learning notebook: inspect the dataset, visualize measurements, and fit a first nearest-neighbor classifier.
For stratified splitting, leakage-safe preprocessing and cross-validation, see the package and walkthrough notebook.

```python
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_lrn, X_test, y_lrn, y_test = train_test_split(iris['data'], iris['target'], random_state=0)
```

```python
print(X_lrn.shape)
print(y_lrn.shape)
```

```python
print(X_test.shape)
print(y_test.shape)
```

```python
import pandas as pd
```

```python
iris_df = pd.DataFrame(X_lrn, columns = iris.feature_names)
```

```python
from pandas.plotting import scatter_matrix
```

```python
%matplotlib inline
```

```python
scatter_matrix(iris_df, c = y_lrn, figsize = (15,15), marker = 'o', hist_kwds = {'bins' : 20}, s = 80, alpha = 0.8)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier(n_neighbors = 1)
```

```python
knn.fit(X_lrn, y_lrn)
```

```python
import numpy as np
X_new = np.array([[5, 2.9, 1, 0.2]])
X_new.shape
```

```python
predict1 = knn.predict(X_new)
print('Prediction class: ', predict1)
print('Prediction type: ', iris['target_names'][predict1])
```

```python
y_predict = knn.predict(X_test)
print(y_predict)
```

```python
print(np.mean(y_predict == y_test))
```

```python
print(knn.score(X_test, y_test))
```
