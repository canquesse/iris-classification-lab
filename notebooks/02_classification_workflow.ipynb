# Iris classification: from measurements to evaluation

A small end-to-end supervised learning workflow. Run cells from top to bottom after installing the project.

```python
from iris_lab.experiment import split_data, run_experiment
import pandas as pd
from pathlib import Path

iris, train, test = split_data(seed=42)
frame = pd.DataFrame(iris.data[train], columns=iris.feature_names)
frame.describe()
```

## Hold out the test set
120 training samples and 30 test samples. Stratification preserves the class balance. Explore only training data before model selection.

```python
print("Training:", len(train), "Test:", len(test))
frame.plot.scatter(x="petal length (cm)", y="petal width (cm)", c=iris.target[train], colormap="viridis")
```

## Compare against a baseline
A majority classifier establishes a reference. Compare logistic regression, nearest neighbors and a small random forest using five stratified folds. Scaling belongs inside each fold. Pick the highest training macro F1 before looking at the test result.

```python
output = Path("../reports") if Path.cwd().name == "notebooks" else Path("reports")
report = run_experiment(output, seed=42)
pd.DataFrame(report["comparison"])
```

```python
print("Selected model:", report["selected_model"])
print("Held-out accuracy:", report["test_accuracy"])
pd.DataFrame(report["classification_report"]).T
```

## Read the errors
The test set is small: one extra error changes accuracy by about 3.3 percentage points. Inspect per-class precision/recall and the confusion matrix. Scores on Iris do not demonstrate real-world deployment readiness.
